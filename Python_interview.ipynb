###Find Out Pairs with given sum in an array in python of time complexity O(n log n)- FACEBOOK,AMAZON

```python

target_sum = 12
arr = [5, 7, 4, 3, 9, 8, 19, 21]
arr.sort()

# Initialize pointers
left = 0
right = len(arr) - 1

# List to store found pairs
found_pairs = []


while left < right:
  current_sum = arr[left] + arr[right]

  if current_sum == target_sum:
    # Found a pair
    found_pairs.append((arr[left], arr[right]))
    left += 1
    right -= 1

  elif current_sum < target_sum:

    left += 1

  else: # current_sum > target_sum

    right -= 1


print(f"pairs found: {found_pairs}")
```

###The Postfix expression to evaluate

```python
# The Postfix expression to evaluate
lst = ["7", "5", "3", "-", "2", "*", "+"]

operator = ['*', '+', '/', '-', '%']

# The 'result' list acts as the evaluation stack
result = []

print(f"Postfix Expression: {lst}")
print("-" * 30)

for x in lst:
  if x not in operator:
    # If the token is an operand (number), push it onto the stack
    result.append(float(x)) # Use float to handle division results

  else:
    # If the token is an operator, pop two operands
    # IMPORTANT: first is the SECOND operand (B)
    first_operand = result.pop()
    # IMPORTANT: second is the FIRST operand (A)
    second_operand = result.pop()

    # Perform the operation: (A operator B)
    if x == '+':
      calculation = second_operand + first_operand
    elif x == '-':
      calculation = second_operand - first_operand # Corrected: A - B
    elif x == '*':
      calculation = second_operand * first_operand
    elif x == '/':
      calculation = second_operand / first_operand # Corrected: A / B
    elif x == '%':
      calculation = second_operand % first_operand # Corrected: A % B

    # Push the result of the calculation back onto the stack
    result.append(calculation)

    # Optional print statement to show the steps:
    # print(f"Operation: {second_operand} {x} {first_operand} = {calculation}. Stack: {result}")

print("-" * 30)
print(f"Final Stack: {result}")
print(f"Evaluation Result: {result[0]}")
```

### N-largest numbers in list- using Min Heap

```python
def sift_down(heap, index, size):
    """Moves an element down the tree until the min-heap property is restored."""
    smallest = index
    left = 2 * index + 1
    right = 2 * index + 2

    # Check if left child exists and is smaller than current
    if left < size and heap[left] < heap[smallest]:
        smallest = left

    # Check if right child exists and is smaller than current
    if right < size and heap[right] < heap[smallest]:
        smallest = right

    if smallest != index:
        heap[index], heap[smallest] = heap[smallest], heap[index]
        sift_down(heap, smallest, size)

def n_largest_heap(arr, n):
    if n <= 0: return []

    # 1. Create a heap from the first n elements
    heap = arr[:n]

    # Build the initial Min-Heap (start from the last non-leaf node)
    for i in range(len(heap) // 2 - 1, -1, -1):
        sift_down(heap, i, n)

    # 2. Process the rest of the array
    for i in range(n, len(arr)):
        if arr[i] > heap[0]:
            heap[0] = arr[i]  # Replace the smallest
            sift_down(heap, 0, n)  # Push the new value to its correct spot

    return heap

arr = [1, 2, 44, 55, 64, 53, 3, 63, 32, 534, 252, 32, 53]
print(n_largest_heap(arr, 3)) # Output: [64, 534, 252] (Order may vary, but values are correct)
```

```python
def reverse_sublist(arr, start, end):
    # Manual swap logic without inbuilt functions
    while start < end:
        arr[start], arr[end] = arr[end], arr[start]
        start += 1
        end -= 1

def rotate_in_place(arr, n, direction="left"):
    l = len(arr)
    if l == 0: return arr
    n = n % l  # Handles n > l

    if direction == "left":
        # Step 1: Reverse 0 to n-1
        reverse_sublist(arr, 0, n - 1)
        # Step 2: Reverse n to end
        reverse_sublist(arr, n, l - 1)
        # Step 3: Reverse the whole thing
        reverse_sublist(arr, 0, l - 1)

    elif direction == "right":
        # Step 1: Reverse the whole thing
        reverse_sublist(arr, 0, l - 1)
        # Step 2: Reverse 0 to n-1
        reverse_sublist(arr, 0, n - 1)
        # Step 3: Reverse n to end
        reverse_sublist(arr, n, l - 1)

# Example usage:
my_arr = [1, 2, 3, 4, 5, 6, 7]
rotate_in_place(my_arr, 2, "left")
print(my_arr) # Output: [3, 4, 5, 6, 7, 1, 2]
```

###all possible words can be generated from

```python
def generate_fixed_length_permutations_generator(s):
    chars = list(s)
    n = len(chars)

    def recurse(current_permutation, remaining_indices):
        # Base case: If the permutation is complete, yield the result
        if len(current_permutation) == n:
            yield "".join(current_permutation)
            return

        # Recursive step:
        for i in range(len(remaining_indices)):
            char_index = remaining_indices[i]
            char_to_add = chars[char_index]
            new_remaining_indices = remaining_indices[:i] + remaining_indices[i+1:]

            # Use 'yield from' to yield results from the recursive calls
            yield from recurse(current_permutation + [char_to_add], new_remaining_indices)

    # Start the recursion and yield the results as they come
    initial_indices = list(range(n))
    yield from recurse([], initial_indices)

input_string_large = "abc"
perms_generator_large = generate_fixed_length_permutations_generator(input_string_large)

count=0
for p in perms_generator_large:
  print(p)
  count+=1
  if "BAPUM"==p:
    break


print(f"after {count} shuffles.")
```

###decimal_to_hexadecial

```python
  # decimal to hexadecimal

def decimal_to_hexadecial(decimal):
  ref={
      10:"A",
      11:"B",
      12:"C",
      13:"D",
      14:"E",
      15:"F"
  }

  bin=[]
  while decimal>0:
    rem=decimal%16
    if rem in ref:
      rem=ref[rem]
    bin.append(str(rem))
    decimal=decimal//16
  return "".join(bin[::-1])

print(decimal_to_hexadecial(203))
```

##Insertion sort


```python
array=[6,8,2,55,1,3,9]

def insertion_sort(array):
  length=len(array)

  # starting anchor at 1 position
  for i in range(1,length):
    temp=array[i] # temp as reference

    j=i-1 # comparing the data left to the temp with temp

    while(j>=0 and array[j]>temp):
      array[j+1]=array[j]
      j-=1
    array[j+1]=temp

  return array

print("before sort:",array)

print("after sort:",insertion_sort(array))
```

###Reverse a string using stack

```python
s = 'abcs'

def reverse_string(s):
    stack = []
    # Push all characters onto the stack
    for c in s:
        stack.append(c)

    new = []
    # Pop characters until the stack is empty
    while len(stack) > 0:
        new.append(stack.pop())

    return "".join(new)

print(reverse_string(s)) # Output: scba
```

###two stacks in array

```python
class TwoStacks:
  def __init__(self, n):
    self.size=n
    self.arr=[0]*n
    self.top1=0
    self.top2=n-1

  def push1(self,val):
    if self.top1 < self.top2-1:
      self.top1+=1
      self.arr[self.top1]=val

    else:
      print("Stack Overflow")

  def push2(self,val):
    if self.top1< self.top2-1:
      self.top2-=1
      self.arr[self.top2]=val

    else:
      print("stack overflow")

  def pop1(self):
    if self.top1>0:
      x= self.arr[self.top1]
      self.top1-=1
      return x
    else:
      print("stack underflow")
      return None


  def pop2(self):
    if self.top2<self.size:
       x= self.arr[self.top2]
       self.top2+=1
       return x
    else:
      print("stack underflow")
      return None


if __name__ == '__main__':
    ts = TwoStacks(5)
    ts.push1(2)
    ts.push1(3)
    ts.push2(4)
    print(ts.pop1(), end=' ')
    print(ts.pop2(), end=' ')
    print(ts.pop2(), end=' ')
```

##Next greater Element-
Monotonic decreasing stack

```python
def next_greater (arr):
  l=len(arr)
  result=[-1]*l
  stk=[]

  #rotation right to left
  for i in range(l-1,-1,-1):
    while stk and arr[stk[-1]]<=arr[i]:
      stk.pop()

    if stk:
            result[i] = arr[stk[-1]]

    # Push the current index onto the stack
    stk.append(i)

  return result

if __name__ == "__main__":

    arr = [6, 8, 0, 1, 3]
    res = next_greater(arr)
    print(" ".join(map(str, res)))
```

###stock pan- monotonic stack application

```python
# stock pan problen
def stack_pan(arr):
  l=len(arr)
  result=[0]*l
  stk=[]

  # right to left rotaion
  for i in range(l):
    freq=1
    #comparing data
    while stk and arr[stk[-1]]<=arr[i]:
      stk.pop()

    # If stack is empty, all elements to the left are smaller
        # Else, top of the stack is the last greater element's index
    if not stk:
        result[i] = (i + 1)
    else:
        result[i] = (i - stk[-1])

    stk.append(i)

  return result

if __name__ == "__main__":

    arr = [10, 4, 5, 90, 120, 80]
    span = stack_pan(arr)

    for x in span:
        print(x, end=" ")
```

###LCM- also using generator if you are iterating over a potentially huge range of numbers and want to keep memory usage low.

```python
from functools import lru_cache

# 1. We wrap the logic in a function to use lru_cache
@lru_cache(maxsize=128)
def find_lcm(n1, n2):
    # 2. Generator expression to yield multiples of the larger number
    # This is memory efficient (O(1) space)
    multiples = (n2 * i for i in range(1, n1 + 1))

    for temp in multiples:
        if temp % n1 == 0:
            return temp

# Usage
num1 = 5
num2 = 133
print(f"Lowest common multiple is {find_lcm(num1, num2)}")
```

```python
d = {"a": 1, "b": {"c": 2, "d": {"e": 3}}}

def flatten_dict(nested_dict, parent_key='', sep='.'):
    items = []
    for k, v in nested_dict.items():
        # Create a new key name (e.g., "b_d_e")
        new_key = f"{parent_key}{sep}{k}" if parent_key else k

        if isinstance(v, dict):
            # If the value is a dict, recurse deeper
            items.extend(flatten_dict(v, new_key, sep=sep).items())
        else:
            # Otherwise, add the value to our list
            items.append((new_key, v))

    return dict(items)

flattened = flatten_dict(d)
print(flattened)
# Output: {'a': 1, 'b_c': 2, 'b_d_e': 3}
```

###Sub arrays - Backtracking

```python
arr = [2, 3, 4, 5, 6, 7, 8, 9, 10, 15]
num = 10

def find_all_subsets(arr, target):
    result = []

    def backtrack(start, current_subset, current_sum):
        # Base Case: If we hit the target, save the subset
        if current_sum == target:
            result.append(list(current_subset))
            return

        # If we exceed the target, stop exploring this path
        if current_sum > target:
            return

        # Explore remaining numbers
        for i in range(start, len(arr)):
            current_subset.append(arr[i])
            # Move to the next index (i + 1) to avoid using the same element twice
            backtrack(i + 1, current_subset, current_sum + arr[i])
            # "Backtrack" by removing the last element added
            current_subset.pop()

    backtrack(0, [], 0)
    return result

print(find_all_subsets(arr, num))
```

```python
import csv

def CSV_reader_chunk(file_name, chunk_size):
    with open(file_name) as file:
        reader = csv.DictReader(file)

        payment_data = {}

        for i, row in enumerate(reader, 1):
            payment = row['payment_method']
            customer = row['customer_name']

            if payment not in payment_data:
                payment_data[payment] = set()

            payment_data[payment].add(customer)

            # yield at chunk boundary
            if i % chunk_size == 0:
                yield {k: len(v) for k, v in payment_data.items()}

        # final leftover
        yield {k: len(v) for k, v in payment_data.items()}

for read_chunck in CSV_reader_chunk('/content/invoices_202510301648.csv', 1000):
  print(read_chunck)
```

separate first_name and last_name- and write it into parquet

```python
import csv
import pyarrow as pa
import pyarrow.parquet as pq

def process_csv_chunk(filepath, chunk_size=1000):
    """Generator that yields processed rows from CSV file in PyArrow Table chunks."""
    with open(filepath, mode='r', encoding='utf-8') as file:
        reader = csv.DictReader(file)
        chunk = []

        for line in reader:
            name_parts = line['customer_name'].split()

            line['first_name'] = name_parts[0]
            line['last_name'] = name_parts[1] if len(name_parts) > 1 else ''

            del line['customer_name'] # Remove original customer_name

            chunk.append(line)
            if len(chunk) >= chunk_size:
                yield pa.Table.from_pylist(chunk)
                chunk = [] # Reset chunk

        # Yield any remaining rows in the last chunk
        if chunk:
            yield pa.Table.from_pylist(chunk)


# --- Main execution --- #
output_filepath = "output.parquet"

# Initialize Parquet writer (if this is the first chunk)
writer = None

for processed_table_chunk in process_csv_chunk("/content/invoices_202510301648.csv", chunk_size=500):
    if writer is None:
        # Create a Parquet writer with the schema from the first chunk
        writer = pq.ParquetWriter(output_filepath, processed_table_chunk.schema)

    # Write the current chunk to the Parquet file
    writer.write_table(processed_table_chunk)

# Close the writer when all chunks have been processed
if writer is not None:
    writer.close()

print("Parquet file written successfully using a generator!")
```

```python
def longest_common_prefix(strs):
    if not strs: # Handle empty list
        return ""

    # Sort the array to easily compare the first and last strings
    strs.sort()
    first_str = strs[0]
    last_str = strs[-1]

    prefix = []
    for i in range(min(len(first_str), len(last_str))):
        if first_str[i] == last_str[i]:
            prefix.append(first_str[i])
        else:
            break
    return "".join(prefix)

# Test cases
words1 = ["flower", "flow", "flight"]
print(f"Longest common prefix for {words1}: {longest_common_prefix(words1)}")

```

###Longest Substring Without Repeating Chars

```python
word = "aqtthbsffasafasfwrwafda"

def get_longest_unique_substring(s):
    char_map = {}  # Stores the last seen index of each character
    left = 0       # Left boundary of the window
    max_len = 0
    start_idx = 0  # Tracking where the longest substring begins

    for right in range(len(s)):
        char = s[right]

        # If we've seen the character and it's inside our current window
        if char in char_map and char_map[char] >= left:
            # Jump the left pointer past the previous occurrence
            left = char_map[char] + 1

        # Update the character's last seen position
        char_map[char] = right

        # Check if the current window is the longest found so far
        if (right - left + 1) > max_len:
            max_len = right - left + 1
            start_idx = left

    return s[start_idx : start_idx + max_len]

print(get_longest_unique_substring(word))
# Output: thbsf
```

### sort 0s, 1s, 2s Dutch National flag approach

```python
arr = [2, 0, 2, 1, 1, 0]

low = 0
mid = 0
high = len(arr) - 1

while mid <= high:
    if arr[mid] == 0:
        # Swap 0 to the front
        arr[low], arr[mid] = arr[mid], arr[low]
        low += 1
        mid += 1
    elif arr[mid] == 1:
        # 1 is in the right place, just move mid
        mid += 1
    else: # arr[mid] == 2
        # Swap 2 to the back
        arr[mid], arr[high] = arr[high], arr[mid]
        high -= 1

print(arr)
# Output: [0, 0, 1, 1, 2, 2]
```
